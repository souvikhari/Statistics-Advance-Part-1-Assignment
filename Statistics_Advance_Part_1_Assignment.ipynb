1. What is a random variable in probability theory?
   - In probability theory, a random variable is a function that assigns a numerical value to each outcome in a sample space of a random experiment.

2. What are the types of random variables?

3. What is the difference between discrete and continuous distributions?
   - The key difference between discrete and continuous distributions lies in the nature of the values they can take. Discrete distributions deal with countable or finite values, while continuous distributions handle values within a range, often an infinite range.

4. What are probability distribution functions (PDF)?
  - In probability theory, a Probability Distribution Function (PDF) describes how the probability is distributed over the values of a random variable. The type of PDF depends on whether the variable is discrete or continuous.

5. How do cumulative distribution functions (CDF) differ from probability distribution functions (PDF)?
   - CDF shows the probability of a random variable being less than or equal to a specific value, while a PDF shows the probability density at a specific point.

6. What is a discrete uniform distribution?
   - A discrete uniform distribution is a type of discrete probability distribution where all outcomes are equally likely.

   - What are the key properties of a Bernoulli distributionKey properties include a binary nature, meaning the random variable can only take the values 0 or 1, and the sum of probabilities for these two outcomes must equal 1.

8. What is the binomial distribution, and how is it used in probability?
   - The binomial distribution is a probability distribution that models the number of successes in a fixed number of independent trials, each with the same probability of success. It's used to calculate the probability of obtaining a certain number of successes when there are only two possible outcomes for each trial, such as success or failure.

9. What is the Poisson distribution and where is it applied?
   - The Poisson distribution is a discrete probability distribution that describes the probability of a certain number of events happening within a specific time or space interval. It's used when events occur independently and at a constant rate, like the number of phone calls arriving at a call center per hour.

10. What is a continuous uniform distribution?
    - A continuous uniform distribution is a type of probability distribution where a continuous random variable is equally likely to take any value within a specific interval.

11. What are the characteristics of a normal distribution?
    - A normal distribution, also known as a Gaussian distribution, is characterized by its bell-shaped, symmetrical curve where the mean, median, and mode are all equal. It's defined by its mean (average) and standard deviation, which determines the spread of the data.

12. What is the standard normal distribution, and why is it important?
    - The standard normal distribution is a specific form of the normal distribution with a mean of 0 and a standard deviation of 1. It's crucial in statistics because it simplifies probability calculations and comparisons across datasets, serving as a reference for any normal distribution.

13. What is the Central Limit Theorem (CLT), and why is it critical in statistics?
    - The Central Limit Theorem (CLT) is a foundational statistical principle stating that the distribution of sample means will approximate a normal distribution, regardless of the original population distribution, as long as the sample size is sufficiently large. This is crucial in statistics because it allows for the application of normal distribution-based methods to a wide range of data, even when the underlying population is not normally distributed.

14. How does the Central Limit Theorem relate to the normal distribution?
    - The Central Limit Theorem bridges non-normal data and normal-based statistical methods by ensuring the sample mean becomes approximately normal as sample size increases.

15. What is the application of Z statistics in hypothesis testing?
    - In hypothesis testing, Z-statistics (or Z-scores) are used to determine how far a sample statistic is from the population parameter, in terms of standard deviations, under the assumption that the null hypothesis is true.

16. How do you calculate a Z-score, and what does it represent?
    - A Z-score represents how many standard deviations a particular data point is away from the mean (average) of a dataset. To calculate a Z-score, you subtract the mean from the individual data point and then divide the result by the standard deviation

17. What are point estimates and interval estimates in statistics?
    - In statistics, a point estimate is a single value used to estimate an unknown population parameter, while an interval estimate provides a range of values within which the true population parameter is likely to fall.

18. What is the significance of confidence intervals in statistical analysis?
    - Confidence intervals in statistical analysis are crucial for understanding the reliability and precision of estimates derived from sample data. They provide a range of values within which the true population parameter is likely to fall, along with a level of confidence in that estimate. This allows for a more nuanced interpretation of results compared to relying solely on point estimates or p-values.

19. What is the relationship between a Z-score and a confidence interval?
    - A Z-score and confidence interval are related because the Z-score is used to determine the margin of error, which is then used to calculate the boundaries of a confidence interval.

20. How are Z-scores used to compare different distributions?
    - Z-scores are used to compare different distributions by standardizing data values, allowing for a consistent measure of how far a data point is from the mean. This standardization enables meaningful comparisons between distributions with potentially different means and standard deviations.

2. Generate a discrete uniform distribution using Python and plot the probability mass function (PMF).

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint

# Parameters for the discrete uniform distribution
low = 1   # inclusive lower bound
high = 7  # exclusive upper bound (like a die roll: 1 through 6)

# Define the distribution
dist = randint(low, high)

# Generate x values (support of the distribution)
x = np.arange(low, high)
pmf = dist.pmf(x)

# Plot the PMF
plt.figure(figsize=(8, 5))
plt.stem(x, pmf, use_line_collection=True)
plt.title('PMF of Discrete Uniform Distribution (randint)')
plt.xlabel('x')
plt.ylabel('P(X = x)')
plt.xticks(x)
plt.ylim(0, 1.1 * max(pmf))
plt.grid(True)
plt.show()

```

3. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution.

```python
def bernoulli_pmf(x, p):
    """
    Compute the PMF of a Bernoulli distribution.

    Parameters:
    x (int): The outcome (0 or 1).
    p (float): The probability of success (P(X=1)), 0 <= p <= 1.

    Returns:
    float: Probability P(X = x)
    """
    if x not in [0, 1]:
        return 0.0  # Bernoulli is only defined for x in {0,1}
    if not (0 <= p <= 1):
        raise ValueError("Probability p must be between 0 and 1.")
    return p**x * (1 - p)**(1 - x)
```

9. Write a Python function to calculate Z-scores from a dataset and plot them.

```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_and_plot_z_scores(data):
    """
    Calculate Z-scores from a dataset and plot them.

    Parameters:
    data (array-like): A list or NumPy array of numerical values.

    Returns:
    z_scores (np.ndarray): Array of Z-scores corresponding to the input data.
    """
    data = np.array(data)
    mean = np.mean(data)
    std = np.std(data)

    if std == 0:
        raise ValueError("Standard deviation is zero. Z-scores cannot be computed.")

    z_scores = (data - mean) / std

    # Plotting Z-scores
    plt.figure(figsize=(10, 5))
    plt.plot(z_scores, marker='o', linestyle='-', color='blue')
    plt.axhline(0, color='black', linewidth=1, linestyle='--', label='Mean (Z=0)')
    plt.axhline(1, color='red', linestyle='--', label='1 Std Dev')
    plt.axhline(-1, color='red', linestyle='--')
    plt.title('Z-Scores of Dataset')
    plt.xlabel('Index')
    plt.ylabel('Z-Score')
    plt.grid(True)
    plt.legend()
    plt.show()

    return z_scores
```

15. Implement hypothesis testing using Z-statistics for a sample dataset.

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def z_test(sample, population_mean, population_std, alpha=0.05):
    """
    Perform a Z-test for a sample dataset to test if the sample mean differs from the population mean.

    Parameters:
    sample (array-like): Sample dataset (list or NumPy array).
    population_mean (float): Known population mean to compare with.
    population_std (float): Known population standard deviation.
    alpha (float): Significance level (default is 0.05).

    Returns:
    z_stat (float): The computed Z-statistic.
    p_value (float): The p-value associated with the Z-statistic.
    """
    # Sample statistics
    sample_mean = np.mean(sample)
    sample_size = len(sample)

    # Calculate Z-statistic
    z_stat = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    # Calculate the p-value (two-tailed test)
    p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))  # two-tailed

    # Decision based on p-value
    if p_value < alpha:
        result = "Reject the null hypothesis: Sample mean is significantly different from population mean."
    else:
        result = "Fail to reject the null hypothesis: No significant difference between sample mean and population mean."

    # Print results
    print(f"Z-statistic: {z_stat:.3f}")
    print(f"P-value: {p_value:.4f}")
    print(result)

    return z_stat, p_value

# Sample data (e.g., exam scores)
sample_data = np.array([55, 58, 59, 63, 60, 62, 55, 54, 58, 60])

# Population parameters
population_mean = 60
population_std = 10

# Perform Z-test
z_stat, p_value = z_test(sample_data, population_mean, population_std)

# Visualizing the result
x = np.linspace(-4, 4, 1000)
pdf = stats.norm.pdf(x, 0, 1)

plt.figure(figsize=(8, 5))
plt.plot(x, pdf, label="Standard Normal Distribution", color='blue')
plt.axvline(z_stat, color='red', linestyle='--', label=f'Z-statistic = {z_stat:.2f}')
plt.fill_between(x, pdf, where=(x > z_stat), color='red', alpha=0.3, label="Rejection Region")
plt.title("Z-Test Visualization")
plt.xlabel("Z-Score")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

16. Create a confidence interval for a dataset using Python and interpret the result.

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

def calculate_confidence_interval(sample_data, confidence_level=0.95):
    """
    Calculate the confidence interval for the mean of a sample dataset.

    Parameters:
    sample_data (array-like): Sample dataset (list or NumPy array).
    confidence_level (float): Desired confidence level (default is 95%).

    Returns:
    tuple: Lower and upper bounds of the confidence interval.
    """
    # Sample statistics
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)
    sample_std = np.std(sample_data, ddof=1)  # Sample standard deviation (ddof=1 for unbiased estimate)

    # Calculate the t-score for the desired confidence level
    alpha = 1 - confidence_level
    t_critical = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)  # Two-tailed t-distribution

    # Calculate the margin of error
    margin_of_error = t_critical * (sample_std / np.sqrt(sample_size))

    # Confidence interval
    ci_lower = sample_mean - margin_of_error
    ci_upper = sample_mean + margin_of_error

    return ci_lower, ci_upper, sample_mean, margin_of_error

# Sample data (e.g., test scores)
sample_data = np.array([55, 58, 59, 63, 60, 62, 55, 54, 58, 60])

# Calculate the confidence interval
ci_lower, ci_upper, sample_mean, margin_of_error = calculate_confidence_interval(sample_data, confidence_level=0.95)

# Print results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"Margin of Error: {margin_of_error:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")

# Plotting the result
plt.figure(figsize=(8, 5))
plt.hist(sample_data, bins=10, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='--', label=f'Mean: {sample_mean:.2f}')
plt.axvline(ci_lower, color='green', linestyle='--', label=f'CI Lower: {ci_lower:.2f}')
plt.axvline(ci_upper, color='green', linestyle='--', label=f'CI Upper: {ci_upper:.2f}')
plt.title('Confidence Interval for Sample Mean')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True)
plt.show()
```

18. Write a Python script to calculate and visualize the probability density function (PDF) of a normal distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def plot_normal_pdf(mean, std_dev):
    """
    Calculate and visualize the Probability Density Function (PDF) of a normal distribution.

    Parameters:
    mean (float): Mean of the normal distribution.
    std_dev (float): Standard deviation of the normal distribution.
    """
    # Define the range of x values (from mean - 4*std_dev to mean + 4*std_dev)
    x = np.linspace(mean - 4*std_dev, mean + 4*std_dev, 1000)

    # Calculate the corresponding PDF values
    pdf_values = stats.norm.pdf(x, loc=mean, scale=std_dev)

    # Plotting the PDF
    plt.figure(figsize=(8, 5))
    plt.plot(x, pdf_values, label=f'Normal Distribution (mean={mean}, std={std_dev})', color='blue')
    plt.fill_between(x, pdf_values, color='skyblue', alpha=0.4)  # Optional: shaded area under the curve
    plt.title('Probability Density Function (PDF) of a Normal Distribution')
    plt.xlabel('x')
    plt.ylabel('Probability Density')
    plt.legend()
    plt.grid(True)
    plt.show()

# Parameters for the normal distribution
mean = 0      # Mean of the distribution
std_dev = 1   # Standard deviation of the distribution

# Call the function to plot the PDF
plot_normal_pdf(mean, std_dev)
```

19.  Use Python to calculate and interpret the cumulative distribution function (CDF) of a Poisson distribution.

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def plot_poisson_cdf(lmbda, max_k):
    """
    Plot the Cumulative Distribution Function (CDF) for a Poisson distribution.

    Parameters:
    lmbda (float): The rate parameter (mean) of the Poisson distribution.
    max_k (int): The maximum value of k for which to calculate and visualize the CDF.
    """
    # Values of k from 0 to max_k
    k_values = np.arange(0, max_k + 1)

    # Calculate the CDF values for the Poisson distribution
    cdf_values = stats.poisson.cdf(k_values, mu=lmbda)

    # Plotting the CDF
    plt.figure(figsize=(8, 5))
    plt.step(k_values, cdf_values, where='post', label=f'Poisson CDF (λ={lmbda})', color='blue', linewidth=2)
    plt.title(f'Cumulative Distribution Function (CDF) of Poisson Distribution (λ={lmbda})')
    plt.xlabel('k (Number of Events)')
    plt.ylabel('CDF (P(X ≤ k))')
    plt.grid(True)
    plt.legend()
    plt.show()

# Parameters for the Poisson distribution
lambda_value = 4  # Rate (mean) of the Poisson distribution
max_k = 15        # Maximum value of k to plot

# Call the function to plot the CDF of the Poisson distribution
plot_poisson_cdf(lambda_value, max_k)
```

20. Simulate a random variable using a continuous uniform distribution and calculate its expected value.

```python
import numpy as np

def simulate_uniform_distribution(a, b, size):
    """
    Simulate a random variable using a continuous uniform distribution.

    Parameters:
    a (float): Lower bound of the uniform distribution.
    b (float): Upper bound of the uniform distribution.
    size (int): Number of random variables to generate.

    Returns:
    np.ndarray: Array of simulated random variables.
    """
    # Simulate random variables from the continuous uniform distribution
    random_variables = np.random.uniform(a, b, size)

    return random_variables

def calculate_expected_value(a, b):
    """
    Calculate the expected value (mean) of a continuous uniform distribution U(a, b).

    Parameters:
    a (float): Lower bound of the uniform distribution.
    b (float): Upper bound of the uniform distribution.

    Returns:
    float: Expected value (mean) of the uniform distribution.
    """
    return (a + b) / 2

# Parameters for the uniform distribution
a = 10  # Lower bound
b = 20  # Upper bound
size = 10000  # Number of random variables to generate

# Simulate random variables
simulated_data = simulate_uniform_distribution(a, b, size)

# Calculate the expected value
expected_value = calculate_expected_value(a, b)

# Calculate the sample mean of the simulated data
sample_mean = np.mean(simulated_data)

# Print the results
print(f"Expected value (theoretical): {expected_value}")
print(f"Sample mean (from simulated data): {sample_mean}")

# Optional: Visualize the simulated data
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(simulated_data, bins=50, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(expected_value, color='red', linestyle='dashed', label=f'Expected value: {expected_value}')
plt.title('Simulation of a Continuous Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.grid(True)
plt.show()
```

21. Write a Python program to compare the standard deviations of two datasets and visualize the difference.

```python
import numpy as np
import matplotlib.pyplot as plt

def compare_standard_deviations(data1, data2):
    """
    Compare the standard deviations of two datasets and visualize the difference.

    Parameters:
    data1 (array-like): First dataset.
    data2 (array-like): Second dataset.
    """
    # Calculate the standard deviation of each dataset
    std_dev_1 = np.std(data1, ddof=1)  # Sample standard deviation (ddof=1 for unbiased estimate)
    std_dev_2 = np.std(data2, ddof=1)

    # Print the standard deviations
    print(f"Standard Deviation of Dataset 1: {std_dev_1:.2f}")
    print(f"Standard Deviation of Dataset 2: {std_dev_2:.2f}")

    # Visualization: Bar Plot of Standard Deviations
    plt.figure(figsize=(8, 5))
    plt.bar(['Dataset 1', 'Dataset 2'], [std_dev_1, std_dev_2], color=['blue', 'green'], alpha=0.7)
    plt.title('Comparison of Standard Deviations of Two Datasets')
    plt.ylabel('Standard Deviation')
    plt.xlabel('Datasets')
    plt.ylim(0, max(std_dev_1, std_dev_2) + 1)  # Set y-axis limit for better visualization
    plt.grid(True, axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Example Datasets
np.random.seed(42)  # For reproducibility
dataset_1 = np.random.normal(50, 10, 1000)  # Dataset 1: Normal distribution (mean=50, std=10)
dataset_2 = np.random.normal(50, 20, 1000)  # Dataset 2: Normal distribution (mean=50, std=20)

# Compare the standard deviations and visualize
compare_standard_deviations(dataset_1, dataset_2)
```

22. Calculate the range and interquartile range (IQR) of a dataset generated from a normal distribution.

```python
import numpy as np

def calculate_range_and_iqr(data):
    """
    Calculate the range and interquartile range (IQR) of a dataset.

    Parameters:
    data (array-like): Input dataset (numeric).

    Returns:
    tuple: (range, interquartile range)
    """
    # Calculate the range (Max - Min)
    data_range = np.max(data) - np.min(data)

    # Calculate the IQR (Q3 - Q1)
    Q1 = np.percentile(data, 25)  # 25th percentile
    Q3 = np.percentile(data, 75)  # 75th percentile
    iqr = Q3 - Q1

    return data_range, iqr

# Generate a dataset from a normal distribution
np.random.seed(42)  # For reproducibility
mean = 50
std_dev = 10
size = 1000  # Number of samples
data = np.random.normal(mean, std_dev, size)

# Calculate range and IQR
data_range, iqr = calculate_range_and_iqr(data)

# Print the results
print(f"Range of the dataset: {data_range:.2f}")
print(f"Interquartile Range (IQR) of the dataset: {iqr:.2f}")
```

23.  Implement Z-score normalization on a dataset and visualize its transformation.

```python
import numpy as np
import matplotlib.pyplot as plt

# Function to normalize data using Z-score
def z_score_normalization(data):
    """
    Perform Z-score normalization (standardization) on the dataset.

    Parameters:
    data (array-like): The input dataset (numeric).

    Returns:
    np.ndarray: The Z-score normalized data.
    """
    mean = np.mean(data)
    std_dev = np.std(data)
    z_scores = (data - mean) / std_dev
    return z_scores

# Generate a dataset from a normal distribution
np.random.seed(42)  # For reproducibility
mean = 50
std_dev = 10
size = 1000  # Number of samples
data = np.random.normal(mean, std_dev, size)

# Apply Z-score normalization
normalized_data = z_score_normalization(data)

# Plotting the original and normalized data
plt.figure(figsize=(12, 6))

# Original data histogram
plt.subplot(1, 2, 1)
plt.hist(data, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.title('Original Data (Before Z-score Normalization)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)

# Normalized data histogram
plt.subplot(1, 2, 2)
plt.hist(normalized_data, bins=30, color='orange', edgecolor='black', alpha=0.7)
plt.title('Normalized Data (After Z-score Normalization)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)

plt.tight_layout()
plt.show()

# Print the mean and standard deviation of both datasets
print(f"Original data: Mean = {np.mean(data):.2f}, Standard Deviation = {np.std(data):.2f}")
print(f"Normalized data: Mean = {np.mean(normalized_data):.2f}, Standard Deviation = {np.std(normalized_data):.2f}")
```

24. Write a Python function to calculate the skewness and kurtosis of a dataset generated from a normal
distribution.

```python
import numpy as np
from scipy.stats import skew, kurtosis

def calculate_skewness_kurtosis(data):
    """
    Calculate the skewness and kurtosis of a dataset.

    Parameters:
    data (array-like): The input dataset (numeric).

    Returns:
    tuple: (skewness, kurtosis)
    """
    # Calculate skewness and kurtosis using scipy.stats
    data_skewness = skew(data)
    data_kurtosis = kurtosis(data)  # Excess kurtosis (kurtosis - 3)

    return data_skewness, data_kurtosis

# Generate a dataset from a normal distribution
np.random.seed(42)  # For reproducibility
mean = 50
std_dev = 10
size = 1000  # Number of samples
data = np.random.normal(mean, std_dev, size)

# Calculate skewness and kurtosis
skewness, kurt = calculate_skewness_kurtosis(data)

# Print the results
print(f"Skewness of the dataset: {skewness:.3f}")
print(f"Kurtosis of the dataset: {kurt:.3f}")
```
